### 1.1 Anomalía

El dataset HIGGS proviene de simulaciones de colisiones de partículas de alta energía en el Gran Colisionador de Hadrones (LHC). En estos experimentos, billones de protones chocan a velocidades cercanas a la de la luz para recrear condiciones extremas del universo. 

El objetivo principal es detectar el **Bosón de Higgs**, una partícula elemental que valida el Modelo Estándar de la física pero que tiene un tiempo de vida demasiado corto antes de desintegrarse en otras partículas (como leptones y jets). El reto radica en que los detectores registran una cantidad colosal de eventos ordinarios producto de interacciones físicas comunes y conocidas. Por lo tanto, el descubrimiento de la partícula de interés equivale a buscar una aguja en un pajar de dimensiones astronómicas.

---

#### Definición Formal de la Anomalía en el Dataset
Dentro del marco analítico de este caso de estudio, se establece que **la Anomalía es el evento de Señal (`is_signal = 1`)**, el cual representa la producción y desintegración del Bosón de Higgs. Por el contrario, los eventos de **Background (`is_signal = 0`)** representan el ruido de fondo ordinario. 

Para abordar metodológicamente este problema, separamos su significado en dos dimensiones técnicas:

#### 1.1.1 Definición Operativa
En la física de partículas experimental, las colisiones que generan partículas de interés son fenómenos sumamente raros y difíciles de replicar. La inmensa mayoría de los eventos registrados corresponden a procesos físicos comunes y ya conocidos, agrupados bajo la etiqueta de Background. Al ser la señal el fenómeno de interés científico pero críticamente escaso, el problema se modela bajo el paradigma operativo de detección de anomalías.

#### 1.1.2. Definición Metodológica y Formulación de la Hipótesis
Desde el punto de vista analítico, evaluaremos la anomalía bajo dos enfoques algorítmicos opuestos:

1. **Enfoque Supervisado (Anomalía por Desbalance Extremo):** La anomalía se aborda puramente por su baja frecuencia (~1.00% de probabilidad de ocurrencia). El reto matemático consiste en mitigar el sesgo del clasificador hacia la clase mayoritaria (Background), forzando al algoritmo a identificar patrones multidimensionales sutiles en los datos a pesar de la escasez de ejemplos.
2. **Enfoque No Supervisado (La Anomalía como Outlier de Densidad):** Bajo este marco teórico, una anomalía se define tradicionalmente como un vector cuyas propiedades numéricas se desvían de la distribución central o habitan en regiones de baja densidad estadística. 

> **Hipótesis Central del Proyecto:** *"Que un evento sea operativamente poco frecuente (Señal) no implica que sea estadísticamente atípico (Outlier)."* 
>
> A través del Análisis Exploratorio de Datos (EDA) desafiaremos esta definición tradicional. Evaluaremos si el Bosón de Higgs realmente habita en las colas aisladas de la distribución (lo que permitiría a modelos no supervisados como `Isolation Forest` aislarlo de forma natural) o si, por el contrario, comparte exactamente el mismo espacio geométrico y rango que el ruido de fondo, diferenciándose únicamente en su nivel de concentración. Esta validación geométrica evalua la superioridad de un enfoque sobre el otro.

---

#### Justificación
Clasificar la señal como una "anomalía" se justifica bajo estos criterios:

* **Justificación por Escasez Operativa:** Los datos confirman que de cada 100,000 eventos, únicamente 1,000 corresponden a la señal (~1.00%). Al estar confinada a esta probabilidad analítica, cumple con la condición de escasez severa requerida para el diseño de sistemas de alerta.
* **Justificación Algorítmica (Asimetría del Error):** Si el modelo confunde el ruido con una señal (Falso Positivo), genera una alerta analizable. Pero si el modelo clasifica una señal como ruido (Falsos Negativo), el Bosón de Higgs se "borra" y se pierde el hito científico. Al requerir que el sistema penalice severamente la pérdida de eventos escasos, el problema se aleja de la clasificación tradicional y se convierte en un diseño de detección y priorización de anomalías.

### 1.3 Señal vs. Background

En esta sección realizamos un análisis estadístico y visual comparativo para identificar si las variables predictoras muestran comportamientos marcadamente distintos cuando el evento es una Señal (1) frente a cuando es Background (0).

Si las medias y las distribuciones de ambos grupos son idénticas, los modelos tendrán serias dificultades para separarlos. Si encontramos variables donde las curvas de densidad no se solapan, habremos descubierto los "puntos clave" que los algoritmos utilizarán para identificar el Bosón de Higgs. Además, esto nos permitirá comprobar si la señal se comporta como un outlier tradicional o no.


```python
df_sin_id = df.drop(columns=['event_id'])
medias_por_clase = df_sin_id.groupby('is_signal').mean().T

medias_por_clase.columns = ['Media_Background_0', 'Media_Señal_1']

medias_por_clase['Diferencia_Absoluta'] = (medias_por_clase['Media_Señal_1'] - medias_por_clase['Media_Background_0']).abs()
medias_por_clase['Cambio_Porcentual (%)'] = (medias_por_clase['Diferencia_Absoluta'] / medias_por_clase['Media_Background_0']) * 100

print("--- VARIABLES CON MAYOR DIFERENCIA DE MEDIAS ---")
medias_por_clase.sort_values(by='Cambio_Porcentual (%)', ascending=False).head(10)
```

Al analizar el cambio porcentual, observamos un fenómeno matemático engañoso en variables como `lepton_eta` (82,872%) y `jet_3_phi` (3,841%). 

En física de partículas, los ángulos de salida (`eta` y `phi`) se distribuyen simétricamente alrededor de cero. Por lo tanto, el promedio real de estas variables tanto para la Señal como para el Background es prácticamente **cero** (por ejemplo, `lepton_eta` en Background es `0.000030`). Al calcular un cambio porcentual dividiendo entre un número tan cercano a cero, cualquier pequeña variación matemática (como `-0.024`) genera un porcentaje astronómico. Físicamente, esto no significa que la señal sea drásticamente diferente, sino que ambas clases están centradas en el origen.

En la separación se observa en variables físicas de masa y energía:

Esto demuestra empíricamente que **la Señal NO es un outlier estadístico**; un leptón proveniente del Bosón de Higgs se mueve con impulsos idénticos a los del ruido común.

Dado que las distribuciones comparten exactamente el mismo espacio geométrico (mismos centros y colas) y solo difieren en su nivel de concentración, confirmamos que **la Señal se esconde de forma idéntica dentro del ruido**. 

Esto tiene una implicación para el caso de estudio: los algoritmos **No Supervisados** (como *Isolation Forest*), que dependen de encontrar puntos en regiones aisladas o de baja densidad del espacio, van a fracasar casi por completo, ya que la señal esta exactamente en la misma región de alta densidad que el background. Por lo tanto, el enfoque **Supervisado** es ayudará mejor para intentar encontrar patrones o interacciones sutiles entre múltiples variables simultáneamente que permitan romper este solapamiento.

### 1.4 Comparación Bajo Nivel vs. Alto Nivel vs. Conjunto Completo

En esta sección realizamos un Análisis Exploratorio de Datos (EDA) comparativo entre los tres conjuntos de características definidos para el caso de estudio. 

Cada bloque de variables aporta una estructura geométrica distinta al espacio de datos. Comparar sus correlaciones, su dispersión y su nivel de solapamiento matemático nos permitirá diagnosticar, antes de entrenar cualquier modelo, qué tan complejo será el límite de decisión y si la ingeniería de características experta (alto nivel) realmente simplifica la tarea de clasificación frente a los datos de bajo nivel.


#### Análisis de Correlación Promedio


```python
def obtener_correlacion_promedio(df_sub):
    corr = df_sub.corr().abs()
    mask = np.eye(corr.shape[0], dtype=bool)
    return corr.values[~mask].mean()

corr_bajo = obtener_correlacion_promedio(df[features_bajo_nivel])
corr_alto = obtener_correlacion_promedio(df[features_alto_nivel])
corr_completo = obtener_correlacion_promedio(df[features_bajo_nivel + features_alto_nivel])

print(f"Correlación absoluta promedio en Bajo Nivel: {corr_bajo:.4f}")
print(f"Correlación absoluta promedio en Alto Nivel: {corr_alto:.4f}")
print(f"Correlación absoluta promedio en Conjunto Completo: {corr_completo:.4f}\n")
```

El contraste cuantitativo de la correlación absoluta promedio entre los tres bloques de variables nos muestra la naturaleza estructural del dataset:

* **Bajo Nivel (0.0448):** Confirma una independencia casi estadística entre las mediciones físicas en bruto. Esto refleja la naturaleza del detector, donde variables cinemáticas y espaciales independientes (momentos, ángulos) no se condicionan linealmente entre sí. Metodológicamente, esto implica que el patrón de la Señal está disperso en interacciones multivariadas complejas que requieren algoritmos no lineales para ser extraídas.
* **Alto Nivel (0.3236):** Exhibe una conectividad significativamente superior debido a la redundancia matemática. Al ser masas invariantes calculadas mediante la combinación de las variables de bajo nivel, comparten componentes físicos comunes en sus ecuaciones, lo que eleva su correlación. Esto demuestra que el bloque de alto nivel actúa como un espacio de información pre-sintetizada y guiada por el conocimiento del dominio.
* **Conjunto Completo (0.0718):** El promedio global se diluye debido a la dominancia numérica de las 21 variables de bajo nivel sobre las 7 de alto nivel. 

Los modelos basados en árboles de decisión se beneficiarán enormemente de la coexistencia de ambos: utilizarán las variables de alto nivel como "atajos predictivos" gracias a su síntesis matemática, mientras usan las de bajo nivel para afinar la frontera de decisión con los datos puros en bruto.

#### 2.0 Partición estratificada en entrenamiento, validación y prueba

La partición habitual en entrenamiento y prueba resulta
insuficiente, porque el apartado 2.5 exige ajustar el umbral de
decisión utilizando el conjunto de validación. El umbral es un hiperparámetro de operación y se elige
observando métricas. Si se eligiera observando el conjunto de prueba, ese conjunto dejaría de ser una muestra
ciega y las métricas finales quedarían sesgadas al alza.

Por lo tanto, se estratifica la data de la siguiente manera,

- Entrenamiento 60%, Ajustar los parámetros de los modelos y aplicar el remuestreo.
- Validación 20%, Seleccionar el modelo y calibrar el umbral de decisión.
- Prueba 20%, Estimar el desempeño final una única vez, sin retroalimentación.

Se elige así, ya que con una prevalencia del 1 %, una partición aleatoria simple
podría repartir los positivos de forma desigual por puro azar. La estratificación fuerza a que
las tres particiones conserven la misma proporción de señales, garantizando que las métricas sean comparables
entre conjuntos. Con 4000 señales disponibles, la partición produce aproximadamente 2400 / 800 / 800
positivos.

```python
X = df[features_completo].copy()
y = df[objetivo[0]].to_numpy()
ids = df[identificadores[0]].to_numpy()

# Primer corte: se aparta el 20 % como conjunto de prueba definitivo.
X_temp, X_test, y_temp, y_test, id_temp, id_test = train_test_split(
    X, y, ids, test_size=0.20, stratify=y, random_state=RANDOM_STATE)

# Segundo corte: del 80 % se toma el 20 % del total.
X_train, X_val, y_train, y_val, id_train, id_val = train_test_split(
    X_temp, y_temp, id_temp, test_size=0.25, stratify=y_temp, random_state=RANDOM_STATE)

resumen_particion = pd.DataFrame({
    "Conjunto": ["Entrenamiento", "Validacion", "Prueba"],
    "Eventos": [len(y_train), len(y_val), len(y_test)],
    "Senales": [int(y_train.sum()), int(y_val.sum()), int(y_test.sum())],
})
resumen_particion["Background"] = resumen_particion["Eventos"] - resumen_particion["Senales"]
resumen_particion["% del total"] = (100 * resumen_particion["Eventos"] / len(y)).round(2)
resumen_particion["Prevalencia (%)"] = (
    100 * resumen_particion["Senales"] / resumen_particion["Eventos"]).round(4)

display(resumen_particion)
```

#### 2.0.1 Estandarización 

El análisis de dispersión de la Sección 1 concluyó que las variables de bajo nivel presentan aproximadamente
el doble de desviación estándar que las de alto nivel, y que ello justifica aplicar StandardScaler antes
de usar modelos sensibles a la escala. 

La media y la desviación estándar son parámetros que se aprenden de los datos. Si se calcularan sobre el dataset completo, los conjuntos de
validación y prueba habrían influido en la transformación y existiría fuga de información
La estandarización se aplica a la Regresión Logística por dos motivos,

1. Convergencia numérica, El optimizador converge de forma más estable cuando todas las variables
   comparten escala.
2. Comparabilidad de coeficientes, Solo con predictores estandarizados la magnitud de un coeficiente
   refleja la importancia relativa de la variable.

```python
escalador = StandardScaler().fit(X_train)   # Ajuste en entrenamiento

X_train_esc = escalador.transform(X_train)
X_val_esc   = escalador.transform(X_val)
X_test_esc  = escalador.transform(X_test)

verificacion = pd.DataFrame({
    "Media (train escalado)": X_train_esc.mean(axis=0),
    "Desv. (train escalado)": X_train_esc.std(axis=0),
    "Media (val escalado)":   X_val_esc.mean(axis=0),
    "Media (test escalado)":  X_test_esc.mean(axis=0),
}, index=features_completo)


print("El entrenamiento queda centrado exactamente en 0 con desviacion 1.")
print("Validacion y prueba quedan CERCA de 0, pero no exactamente, no participaron en el ajuste.\n")
display(verificacion.head(6).round(6))
```

Funciones reutilizables

```python
def evaluar_umbral(y_true, y_score, umbral, nombre=""):
    """Calcula el conjunto completo de metricas para un umbral de decision dado."""
    y_pred = (y_score >= umbral).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    return {
        "Modelo": nombre,
        "Umbral": float(umbral),
        "AP": average_precision_score(y_true, y_score),
        "ROC-AUC": roc_auc_score(y_true, y_score),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "Alertas": int(y_pred.sum()),
        "TP": int(tp), "FP": int(fp), "FN": int(fn), "TN": int(tn),
    }


def umbral_maximo_f1(y_true, y_score):
    """Devuelve el umbral que maximiza F1. Debe invocarse SOLO sobre validacion."""
    precision, recall, umbrales = precision_recall_curve(y_true, y_score)
    f1 = 2 * precision * recall / (precision + recall + 1e-12)
    return float(umbrales[int(np.nanargmax(f1[:-1]))])


def umbral_por_presupuesto(y_score, n_alertas):
    """Devuelve el umbral que produce exactamente n_alertas eventos priorizados."""
    return float(np.sort(y_score)[::-1][n_alertas - 1])


def metricas_top_k(y_true, y_score, k):
    """Evalua unicamente los k eventos de mayor prioridad (vision operativa)."""
    orden = np.argsort(y_score)[::-1][:k]
    tp = int(np.asarray(y_true)[orden].sum())
    total_positivos = int(np.asarray(y_true).sum())
    return {"k": k, "TP@k": tp, "Precision@k": tp / k,
            "Recall@k": tp / total_positivos,
            "Lift@k": (tp / k) / np.asarray(y_true).mean()}


def graficar_matriz_confusion(y_true, y_score, umbral, titulo, ax=None):
    """Dibuja la matriz de confusion correspondiente a un umbral dado."""
    y_pred = (y_score >= umbral).astype(int)
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    if ax is None:
        _, ax = plt.subplots(figsize=(4, 3.4))
    sns.heatmap(cm, annot=True, fmt=",d", cmap="viridis", cbar=False,
                xticklabels=["Background", "Senal"],
                yticklabels=["Background", "Senal"], ax=ax)
    ax.set_xlabel("Prediccion"); ax.set_ylabel("Valor real"); ax.set_title(titulo, fontsize=10)
    return ax


# Presupuesto operativo de alertas: 1 % del conjunto evaluado.
PRESUPUESTO_TEST = int(round(0.01 * len(y_test)))
PRESUPUESTO_VAL  = int(round(0.01 * len(y_val)))

print("Funciones de evaluacion definidas.")
print(f"Presupuesto de alertas en validacion : {PRESUPUESTO_VAL:,} eventos")
print(f"Presupuesto de alertas en prueba     : {PRESUPUESTO_TEST:,} eventos")
```

### 2.2 Segundo modelo de scikit-learn: HistGradientBoostingClassifier

#### 2.2.1 Justificación de la elección

La elección de
HistGradientBoostingClassifier responde a tres criterios explícitos,

1. Coherencia con el diagnóstico de la Sección 1. El análisis concluyó que el patrón de la señal
   reside en interacciones multivariadas complejas y que los modelos de árboles aprovecharían la
   coexistencia de ambos bloques usando las variables de alto nivel como atajos predictivos. 
2. Capacidad de representación. El boosting construye árboles de forma secuencial, y cada árbol nuevo
   se especializa en corregir los errores del conjunto previo. Los árboles capturan interacciones y
   umbrales no lineales que la frontera lineal del modelo base no puede representar.
3. Viabilidad computacional. La variante histogram-based discretiza cada predictor en 256 grupos, lo
   que reduce drásticamente el costo de evaluar cortes candidatos y permite entrenar sobre 240000
   observaciones de forma eficaz.

```python
hgb_base = HistGradientBoostingClassifier(
    max_iter=400,              # numero de arboles del ensamble
    learning_rate=0.08,        # contribucion de cada arbol, valores bajos exigen mas arboles
    max_leaf_nodes=31,         # complejidad maxima por arbol
    l2_regularization=1.0,     # penalizacion que limita el sobreajuste
    early_stopping=False,     
    random_state=RANDOM_STATE,
)
hgb_base.fit(X_train, y_train)  

prob_val_hgb_base  = hgb_base.predict_proba(X_val)[:, 1]
prob_test_hgb_base = hgb_base.predict_proba(X_test)[:, 1]

ap_val_hgb = average_precision_score(y_val, prob_val_hgb_base)
ap_val_lr  = average_precision_score(y_val, prob_val_lr_base)


print(f"AP  Regresion Logistica  : {ap_val_lr:.4f}")
print(f"AP  HistGradientBoosting : {ap_val_hgb:.4f}")
print(f"Mejora relativa          : {100*(ap_val_hgb-ap_val_lr)/ap_val_lr:+.1f}%")
print(f"\nAP del modelo de arboles sobre el azar: x{ap_val_hgb/y_val.mean():.2f}")
```

Interpretación. La mejora sustancial del modelo de árboles sobre el modelo lineal constituye la

### 2.3 Comparación del modelo original con estrategias de tratamiento del desbalance

El enunciado exige comparar el modelo original con al menos dos estrategias. Se implementan tres,
elegidas porque intervienen en mecanismos distintos del proceso de aprendizaje,

| Estrategia | Mecanismo de acción | Efecto sobre los datos |
|---|---|---|
| Pesos de clase | Modifica la función de pérdida: cada error sobre una señal pesa ~99 veces más | Ninguno: los datos permanecen intactos |
| Random Undersampling | Elimina observaciones de la clase mayoritaria | Reduce drásticamente el volumen de entrenamiento |
| SMOTE | Genera positivos sintéticos interpolando entre vecinos cercanos reales | Aumenta el volumen con datos artificiales |

#### 2.3.1 Justificación de la proporción de remuestreo

Tanto RandomUnderSampler como SMOTE aceptan el parámetro sampling_strategy. Su valor por defecto
('auto') produce un balance perfecto 1:1, opción que se descarta deliberadamente,

- Con undersampling al 1:1 el entrenamiento se reduciría de 240 000 a unas 4 800 observaciones,
  descartando más del 98 % de la información disponible sobre el background. Dado que el hallazgo de
  la Sección 1 es que ambas clases se solapan, caracterizar bien el background resulta esencial.
- Con SMOTE al 1:1 se generarían ~237 000 positivos sintéticos a partir de solo 2400 reales, es decir, unas
  99 copias interpoladas por cada señal auténtica. El modelo aprendería la geometría del interpolador
  más que la física de la señal.

Se adopta sampling_strategy=0.1. Esto eleva la
prevalencia efectiva del 1 % al ~9 %, suficiente para alterar el punto de operación, sin los efectos
adversos del balanceo total.

#### 2.3.2 Diseño experimental de la comparación

Para que la comparación sea controlada, las cuatro variantes se aplican al mismo algoritmo manteniendo idénticos el resto de los elementos, mismos datos, mismo escalador, mismas métricas y
mismo criterio de umbral. Así, cualquier diferencia observada es atribuible únicamente a la estrategia.

Cada variante se evalúa en dos puntos de operación para separar dos efectos que suelen confundirse,

- Umbral fijo 0.5: revela cómo cada estrategia recalibra las probabilidades del modelo.
- Umbral óptimo de F1 calibrado en validación, revela si la estrategia mejora el desempeño alcanzable
  una vez que el umbral se elige correctamente.

```python
def entrenar_variante(X_ajuste, y_ajuste, nombre, class_weight=None):
    """Entrena una Regresion Logistica y evalua en dos puntos de operacion."""
    modelo = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE,
                                class_weight=class_weight)
    modelo.fit(X_ajuste, y_ajuste)

    prob_val  = modelo.predict_proba(X_val_esc)[:, 1]
    prob_test = modelo.predict_proba(X_test_esc)[:, 1]

    umbral_f1 = umbral_maximo_f1(y_val, prob_val)          # calibrado en validación
    res_05 = evaluar_umbral(y_test, prob_test, 0.5, nombre)
    res_f1 = evaluar_umbral(y_test, prob_test, umbral_f1, nombre)

    return {
        "modelo": modelo, "prob_val": prob_val, "prob_test": prob_test,
        "umbral_f1": umbral_f1, "res_05": res_05, "res_f1": res_f1,
        "n_entrenamiento": len(y_ajuste), "n_positivos": int(np.sum(y_ajuste)),
    }


variantes = {}

variantes["1. Sin tratamiento"] = entrenar_variante(
    X_train_esc, y_train, "LR sin tratamiento")

variantes["2. Pesos de clase"] = entrenar_variante(
    X_train_esc, y_train, "LR pesos de clase", class_weight="balanced")

submuestreador = RandomUnderSampler(sampling_strategy=0.1, random_state=RANDOM_STATE)
X_rus, y_rus = submuestreador.fit_resample(X_train_esc, y_train)
variantes["3. Random Undersampling"] = entrenar_variante(X_rus, y_rus, "LR undersampling")

sintetizador = SMOTE(sampling_strategy=0.1, k_neighbors=5, random_state=RANDOM_STATE)
X_smote, y_smote = sintetizador.fit_resample(X_train_esc, y_train)
variantes["4. SMOTE"] = entrenar_variante(X_smote, y_smote, "LR SMOTE")

composicion = pd.DataFrame([
    {"Estrategia": k,
     "Observaciones": v["n_entrenamiento"],
     "Positivos": v["n_positivos"],
     "Prevalencia (%)": round(100 * v["n_positivos"] / v["n_entrenamiento"], 2)}
    for k, v in variantes.items()]).set_index("Estrategia")
display(composicion)
```

#### 2.3.3 Resultados: el efecto real de cada estrategia

La tabla siguiente reúne los dos puntos de operación. La columna clave para la interpretación es AP,
porque es la única métrica que no depende del umbral.

```python
filas = []
for nombre, v in variantes.items():
    filas.append({
        "Estrategia": nombre,
        "AP": v["res_f1"]["AP"],
        "Alertas @0.5": v["res_05"]["Alertas"],
        "Recall @0.5": v["res_05"]["Recall"],
        "F1 @0.5": v["res_05"]["F1"],
        "Umbral optimo": round(v["umbral_f1"], 4),
        "Alertas @optimo": v["res_f1"]["Alertas"],
        "Precision @optimo": v["res_f1"]["Precision"],
        "Recall @optimo": v["res_f1"]["Recall"],
        "F1 @optimo": v["res_f1"]["F1"],
    })

tabla_desbalance = pd.DataFrame(filas).set_index("Estrategia")
display(tabla_desbalance.round(4))

rango_ap = tabla_desbalance["AP"].max() - tabla_desbalance["AP"].min()
print(f"Rango de variacion del AP entre las cuatro estrategias : {rango_ap:.4f}")
print(f"Rango de variacion de las alertas con umbral 0.5       : "
      f"{tabla_desbalance['Alertas @0.5'].min():,} a {tabla_desbalance['Alertas @0.5'].max():,}")
```

#### 2.5.1 El umbral como decisión de negocio

El apartado 2.1.2 evidenció que el umbral 0.5 es inservible en este contexto. La cuestión relevante es
qué criterio debe sustituirlo, y la respuesta depende del objetivo operativo. Se implementan y comparan
dos criterios de naturaleza distinta,

| Criterio | Definición | Cuándo es apropiado |
|---|---|---|
| Máximo F1 | Umbral que maximiza la media armónica de precisión y recall | Cuando ambos errores tienen costos comparables y no hay restricción de capacidad |
| Presupuesto de alertas | Umbral que produce exactamente k alertas (aquí el 1 % superior) | Cuando la capacidad de revisión es limitada, situación habitual en la práctica |

El segundo criterio es el que corresponde a la realidad de un experimento de física de partículas, el equipo
dispone de una capacidad finita de análisis detallado y necesita saber qué eventos revisar primero.

### 3.1.0 Funciones para evaluacion del modelo

- generacion de mascara de alertas, para evaluar distintos valores de contaminacion sobre los scores generados por el modelo
- calculo de metricas recision, recall, F1, numero de alertas y Average Precision 
- definicion de valores de contaminacion a validar

```python
# retorna el porcentaje superior de los scores,
# donde mayor = mas anomalo, basado en un parametro de contaminacion
def alert_mask_scores(scores, contamination):
    umbral = np.quantile(scores, 1 - contamination)
    mask = scores >= umbral

    return mask, umbral


def assess_alerts(y_true, scores, contamination):
    y_true = np.asarray(y_true)

    mask, umbral = alert_mask_scores(scores, contamination)

    alerts = int(mask.sum())

    tn, fp, fn, tp = confusion_matrix(y_true, mask).ravel()

    precision = precision_score(y_true, mask, zero_division=0)
    recall = recall_score(y_true, mask, zero_division=0)
    f1 = f1_score(y_true, mask, zero_division=0)

    ap = average_precision_score(y_true, scores)

    return {
        "contamination": contamination,
        "umbral": umbral,
        "alertas" : alerts,
        "tp": tp, "fp": fp, "fn": fn, "tn": tn,
        "precision": precision, "recall": recall,
        "f1": f1, "ap": ap
    }

def assess_contamination_vals(y_true, scores, cont_vals):
    records = [assess_alerts(y_true, scores, cont) for cont in cont_vals]

    return pd.DataFrame(records)

CONTAMINATION_VALUES = [0.005, 0.01, 0.02, 0.03, 0.05, 0.08, 0.10]
```

### 3.1 Isolation Forest

Isolation Forest se entrena **sin etiquetas**. El modelo se ajusta una sola vez por cada bloque de variables y posteriormente se obtienen los anomaly scores continuos. 
En este análisis, se utiliza -decision_function() para adoptar la convención de que un score mayor representa una mayor probabilidad de anomalía.

El parámetro `contamination` del constructor se utiliza para establecer el umbral interno de clasificación del modelo, 
pero no es necesario volver a entrenar el modelo para explorar distintos niveles de alerta. 
Por ello, una vez obtenido el conjunto de scores, se evalúan diferentes niveles de contaminación directamente sobre estos scores mediante distintos umbrales.

La elección final del nivel de contaminación combina dos fuentes de evidencia:

- **Prior de dominio**: el enunciado del caso establece que el dataset fue preparado para que la señal represente aproximadamente el 1% de los registros. Este valor constituye 
un punto de partida razonable y es información disponible antes de utilizar las etiquetas individuales para evaluar el modelo.


- **Barrido en validación**: manteniendo fijo el modelo entrenado, se revelan las etiquetas del conjunto de validación únicamente para evaluar una grilla de niveles de contaminación. 
  Para cada nivel se calcula el umbral correspondiente y se analizan precision, recall, F1-score y Average Precision, permitiendo estudiar el compromiso entre la detección de anomalías y la generación de falsas alertas.
  El nivel seleccionado se determina utilizando exclusivamente el conjunto de validación.

Al finalizar, el nivel de contaminación seleccionado se mantiene fijo y se aplica una sola vez al conjunto de test, utilizando las etiquetas de test únicamente para la evaluación final del desempeño.

```python
def train_iforest(X_train, X_val, X_test, n_estimators=200):
    model = IsolationForest(
        n_estimators= n_estimators,
        contamination= 0.01,
        max_samples="auto",
        random_state=RANDOM_STATE,
        n_jobs= -1
    )

    model.fit(X_train)
    scores_val = -model.decision_function(X_val)
    scores_test = -model.decision_function(X_test)
    return model, scores_val, scores_test

models_iforest = {}
scores_iforest_val = {}
scores_iforest_test = {}

for name, data in df_features.items():

    train = data['train']
    val = data['val']
    test = data['test']

    model, s_val, s_test = train_iforest(train, val, test)
    
    models_iforest[name] = model
    scores_iforest_val[name] = s_val
    scores_iforest_test[name] = s_test

    print(
        f'Isolation Forest [{name}] entrenado sobre {train.shape[0]:,} observaciones'
        f' y {train.shape[1]} variables.'
        )

```

```python
metrics_iforest = {}

for name in df_features:
    
    metrics = assess_contamination_vals(
        y_val, 
        scores_iforest_val[name], 
        CONTAMINATION_VALUES
    )

    metrics_iforest[name] = metrics

    print(f'\nBarrido de contaminacion(validacion) - IsolationForest[{name}]')
    display(metrics[["contamination", "alertas", "precision", "recall", "f1", "ap"]])
```



### 3.1.2 Selección del nivel de contaminación

Con base en el barrido realizado sobre el conjunto de validación, se selecciona para cada bloque de variables el nivel de `contamination` que obtiene el mejor **F1-score**, considerando un límite máximo de **5% de alertas** (`MAX_ALERT_BUDGET`).

Este límite se establece como un **supuesto operativo**, ya que el enunciado no proporciona información sobre la capacidad real de revisión de alertas. Su objetivo es evitar seleccionar un nivel que genere un volumen excesivo de casos para analizar.

Esta restricción también es relevante porque el **Average Precision** obtenido es cercano a la prevalencia base de la clase (aproximadamente 1%), lo que indica una capacidad limitada del modelo para separar las anomalías de los casos normales. En consecuencia, al aumentar el número de alertas, el *recall* tiende a aumentar y el F1-score puede favorecer niveles altos de contaminación, aunque estos generen demasiadas alertas para ser útiles en la práctica.

Por esta razón, la selección del nivel de contaminación se realiza **dentro del límite máximo del 5%**, buscando el mejor equilibrio entre *precision* y *recall*. El valor seleccionado puede ser diferente del **1% utilizado como prior de dominio**, ya que el 1% representa la prevalencia esperada de la señal, mientras que el barrido busca determinar qué nivel de alertas ofrece el mejor desempeño observado en el conjunto de validación.


```python
MAX_ALERT_BUDGET = 0.05

def choose_contamination(barrido, max_budget=MAX_ALERT_BUDGET):
    potential_contamination = barrido[barrido["contamination"] <= max_budget]
    return potential_contamination.loc[potential_contamination["f1"].idxmax(), "contamination"]


chosen_contamination_iforest = {
    name: choose_contamination(metrics_iforest[name])
    for name in df_features
}
print(f"Contaminacion elegida por bloque (Isolation Forest, tope de alertas={MAX_ALERT_BUDGET}):")
for name, c in chosen_contamination_iforest.items():
    print(f"  {name}: {c}")
```

```python
results_iforest_test = []

for name in df_features:
    cont = chosen_contamination_iforest[name]
    metrics = assess_alerts(y_test, scores_iforest_test[name], cont)

    metrics['modelo'] = "Isolation Forest"
    metrics['bloque'] = name

    results_iforest_test.append(metrics)


df_results_iforest = pd.DataFrame(results_iforest_test)

df_results_iforest[["modelo", "bloque", "contamination", "alertas", "tp", "fp", "fn",
                       "precision", "recall", "f1", "ap"]]
```

Los resultados muestran que el `bloque de bajo nivel` obtiene el mejor desempeño entre las tres configuraciones evaluadas, con un **F1 de 0.01625** y un **Average Precision de 0.00953**. El bloque de variables completas presenta un desempeño ligeramente inferior (F1 = 0.01250, AP = 0.00940), mientras que el bloque de alto nivel obtiene los valores más bajos (F1 = 0.00938, AP = 0.00916).

Sin embargo, las diferencias entre los valores de Average Precision son pequeñas y todos se encuentran alrededor de la prevalencia base de la señal, aproximadamente 1%. Esto indica que **Isolation Forest no consigue establecer un ranking significativamente mejor que el esperado por azar** para ninguno de los bloques de variables.

En el punto de operación seleccionado, el bloque de bajo nivel identifica 39 de las 800 señales reales (recall = 4.88%), pero genera 3,961 falsas alertas de un total de 4,000. Por tanto, aunque presenta el mejor resultado relativo entre las configuraciones, **su capacidad de detección sigue siendo limitada**.

En consecuencia, la hipótesis inicial de que las variables de alto nivel proporcionarían un mejor Average Precision no se confirma con estos resultados. Asimismo, la incorporación de todas las variables no produce una mejora respecto al bloque de bajo nivel.

### 3.2 LOF

- **`n_neighbors = 20`**: valor estandar recomendado en la literatura de LOF; con ~240,000 observaciones de entrenamiento es suficiente para estimar la densidad local sin ser demasiado sensible al ruido de vecinos individuales.
- **Submuestreo del conjunto de entrenamiento**: LOF con `novelty=True` calcula, para cada observacion nueva, sus distancias contra el conjunto de referencia usado en el ajuste. Su costo computacional crece de forma no lineal con el tamano de dicho conjunto. Ajustar sobre las 240,000 observaciones completas es computacionalmente impracticable en un notebook exploratorio; se utiliza una **submuestra aleatoria de 80,000 observaciones de entrenamiento** (sin usar la etiqueta para seleccionarla, por lo que no introduce fuga de informacion) como referencia de densidad local, una practica estandar para escalar metodos basados en vecinos.
- **Contaminacion / umbral**: se aplica exactamente la misma metodologia que en Isolation Forest (prior de dominio + barrido en validacion maximizando F1), para mantener la comparacion entre modelos consistente.

```python
LOF_TRAIN_SUBSAMPLE = 80_000

def train_lof(X_train, X_val, X_test, n_neigbors=20, subsample=LOF_TRAIN_SUBSAMPLE, seed=RANDOM_STATE):
    rng = np.random.default_rng(seed)

    n = X_train.shape[0]

    if subsample < n:
        idx = rng.choice(n, size=subsample, replace=False)
        X_train_ref = X_train[idx]
    else:
        X_train_ref = X_train

    
    model = LocalOutlierFactor(
        n_neighbors=n_neigbors,
        novelty=True,
        contamination=0.01,
        n_jobs=-1
    )

    model.fit(X_train_ref)

    scores_val = -model.decision_function(X_val)
    scores_test = -model.decision_function(X_test)

    return model, scores_val, scores_test

models_lof = {}
scores_lof_val = {}
scores_lof_test = {}

for name, data in df_features.items():
    train = data['train']
    val = data['val']
    test = data['test']

    model, s_val, s_test = train_lof(train, val, test)

    models_lof[name] = model
    scores_lof_val[name] = s_val
    scores_lof_test[name] = s_test

    print(f'LOF [{name}] ajustado con muestra de {min(LOF_TRAIN_SUBSAMPLE, train.shape[0]):,} '
          f'observaciones y {train.shape[1]} variables.')
```

### 4.2. Balance entre Detección de Señales y Falsos Positivos

En esta sección se analiza la relación de intercambio entre las alertas correctas y los errores cometidos por cada algoritmo dentro del presupuesto fijo de 800 alertas en el conjunto de prueba. 

En la operación del laboratorio, cada error tiene un impacto directo:
* **Falso Positivo (FP):** Representa un costo de eficiencia. Implica asignar recursos humanos y tiempo de cómputo a revisar un evento que resulta ser ruido de fondo ordinario.
* **Falso Negativo (FN):** Representa una pérdida científica. Implica que un evento real del Bosón de Higgs fue clasificado como ruido y se descartó del análisis.



```python
analisis_balance = []

for res in resultados_presupuesto:
    tp = res["TP"]
    fp = res["FP"]
    
    fp_por_acierto = fp / tp if tp > 0 else float('inf')
    
    porcentaje_rescate = (tp / PRESUPUESTO_TEST) * 100
    
    analisis_balance.append({
        "Modelo": res["Modelo"],
        "Aciertos (TP)": tp,
        "Falsas Alarmas (FP)": fp,
        "Falsos Positivos por cada Acierto": fp_por_acierto,
        "Señales Rescatadas (%)": porcentaje_rescate
    })

df_balance = pd.DataFrame(analisis_balance)
display(df_balance.round(2))
```

#### 4.2.1. Evaluación del Costo Operativo y Eficiencia

El análisis de la tasa de falsos positivos por cada acierto cuantifica el costo en recursos que exige cada modelo dentro del límite de 800 alertas:

* **Eficiencia de LOF:** El modelo Local Outlier Factor presenta el menor costo de inspección, requiriendo procesar 49.0 eventos de ruido por cada señal correcta detectada. Este enfoque recupera el 2.0% de los eventos del Bosón de Higgs en el conjunto de prueba utilizando solo el 1.0% de la capacidad de revisión disponible.
* **Inadecuación de HGB en el Top-K:** El modelo supervisado eleva el costo operativo a 132.33 falsas alarmas por cada acierto dentro de este segmento prioritario, lo que triplica el esfuerzo de revisión en comparación con LOF y recupera menos del 1.0% de las señales disponibles.

### 4.3. Intersección y Coincidencia de Alertas entre Enfoques

En esta sección se realiza el cruce de las observaciones priorizadas en el conjunto de prueba por el modelo supervisado (HGB) y el modelo no supervisado (LOF). 

A partir de las 800 alertas emitidas por cada algoritmo, los 80,000 registros del conjunto de prueba se clasifican en cuatro categorías exclusivas:
1. **Ambos enfoques:** Eventos prioritarios de coincidencia donde ambos modelos concuerdan en la alerta.
2. **Únicamente Supervisado (HGB):** Casos priorizados exclusivamente por el aprendizaje con etiquetas.
3. **Únicamente No Supervisado (LOF):** Casos identificados solo por su baja densidad estadística local.
4. **Ninguno de los dos:** El resto de las observaciones que no entraron en el presupuesto del 1.00% superior de sospecha.


```python
# Obtener los vectores binarios (True/False) de las 800 alertas de cada modelo
mask_hgb, _ = alert_mask_scores(prob_test_final, 0.01)
mask_lof, _ = alert_mask_scores(scores_lof_test["completo"], 0.01)

# Calculo de las máscaras booleanas para las 4 categorías exclusivas
ambos = mask_hgb & mask_lof
solo_supervisado = mask_hgb & ~mask_lof
solo_no_supervisado = mask_lof & ~mask_hgb
ninguno = ~mask_hgb & ~mask_lof

categorias = ["Ambos Enfoques", "Únicamente Supervisado (HGB)", "Únicamente No Supervisado (LOF)", "Ninguno de los dos"]
mascaras = [ambos, solo_supervisado, solo_no_supervisado, ninguno]

reporte_interseccion = []
for cat, mask in zip(categorias, mascaras):
    total_casos = int(mask.sum())
    senales_reales = int(y_test[mask].sum())
    background_casos = total_casos - senales_reales
    precision_grupo = senales_reales / total_casos if total_casos > 0 else 0.0
    
    reporte_interseccion.append({
        "Categoría de Alerta": cat,
        "Total Observaciones": total_casos,
        "Señales Reales (TP)": senales_reales,
        "Ruido (FP)": background_casos,
        "Precisión en el Grupo (%)": precision_grupo * 100
    })

df_interseccion = pd.DataFrame(reporte_interseccion)
display(df_interseccion.round(4))
```

#### 4.3.1. Diagnóstico de Coincidencia y Ortogonalidad de los Modelos

El cruce de índices de las alertas permite evaluar la complementariedad de los enfoques matemático y supervisado:

Al no existir un solapamiento significativo entre ambos grupos de alertas, se confirma que las observaciones consideradas "anómalas por rareza estadística" difieren casi por completo de las consideradas "anómalas por patrón de señal".
* **Distribución del Residuo:** El grupo clasificado como "Ninguno de los dos" concentra el grueso de las observaciones y de las señales omitidas. Esto reafirma que el presupuesto estricto del 1.00% restringe la capacidad de captura global, forzando a priorizar la precisión local de cada enfoque sobre el volumen total de detecciones.

## 5. Integración de los Enfoques

En esta sección se desarrolla y evalúa una solución integrada que combina las capacidades del enfoque supervisado y del no supervisado. El análisis exploratorio y la confrontación previa demostraron que el modelo supervisado (HGB) y el modelo basado en distancias (LOF) seleccionan observaciones en regiones espaciales distintas y bajo criterios ortogonales.

Se analizará si la combinación de los algoritmos reduce el volumen de falsos positivos y mejora la tasa de captura del Bosón de Higgs frente a los modelos individuales.

### 5.1. Estrategia: Inyección del Puntaje de Anomalía como Característica Predictora

Esta estrategia consiste en incorporar el score continuo de densidad local generado por el modelo `LOF` (conjunto completo) como una nueva variable numérica dentro del espacio de características del modelo supervisado. 

El fundamento técnico es permitir que el algoritmo basado en árboles (`HistGradientBoosting`) aprenda relaciones no lineales combinando las variables físicas originales con la métrica de rareza estadística. Para mantener el rigor metodológico, los scores de LOF se calculan de manera independiente para cada conjunto, evitando la fuga de información hacia el entrenamiento.



```python
scores_lof_train = -models_lof["completo"].decision_function(df_features["completo"]["train"])
scores_lof_val   = scores_lof_val["completo"]
scores_lof_test  = scores_lof_test["completo"]

# Se agrega el score de LOF como una nueva columna al final de las matrices de datos procesados
X_train_ext = np.column_stack((df_features["completo"]["train"], scores_lof_train))
X_val_ext   = np.column_stack((df_features["completo"]["val"], scores_lof_val))
X_test_ext  = np.column_stack((df_features["completo"]["test"], scores_lof_test))

print(f"Dimensiones originales de entrenamiento : {df_features['completo']['train'].shape}")
print(f"Nuevas dimensiones de entrenamiento      : {X_train_ext.shape} (28 variables + Score LOF)")

# Inicializar y reentrenar el HistGradientBoostingClassifier con los mismos hiperparámetros
hgb_integrado = HistGradientBoostingClassifier(
    max_iter=200, learning_rate=0.08, max_leaf_nodes=31,
    l2_regularization=1.0, early_stopping=False, random_state=RANDOM_STATE
)

print("\nEntrenando el modelo supervisado integrado...")
hgb_integrado.fit(X_train_ext, y_train)

prob_val_hgb_ext  = hgb_integrado.predict_proba(X_val_ext)[:, 1]
prob_test_hgb_ext = hgb_integrado.predict_proba(X_test_ext)[:, 1]

umbral_dinamico_ext = umbral_por_presupuesto(prob_test_hgb_ext, PRESUPUESTO_TEST)
res_integrado = evaluar_umbral(y_test, prob_test_hgb_ext, umbral_dinamico_ext, nombre="HGB + Score LOF (Integrado)")
metrics_k_ext = metricas_top_k(y_test, prob_test_hgb_ext, PRESUPUESTO_TEST)

res_integrado["Precision@K"] = metrics_k_ext["Precision@k"]
res_integrado["Recall@K"] = metrics_k_ext["Recall@k"]
res_integrado["Lift@K"] = metrics_k_ext["Lift@k"]

# Recuperar el resultado previo del HGB individual para comparar directamente
res_individual = df_comparativa_alertas[df_comparativa_alertas["Modelo"] == "Supervisado (HGB)"].iloc[0].to_dict()

df_comparativa_estrategia_a = pd.DataFrame([res_individual, res_integrado])
display(df_comparativa_estrategia_a[["Modelo", "Alertas", "TP", "FP", "Precision@K", "Recall@K", "Lift@K", "AP"]].round(4))
```

#### 5.1.1. Diagnóstico de la Estratificación de la Estrategia

El reentrenamiento del modelo supervisado incorporando el score continuo de densidad local de LOF genera un incremento drástico en la capacidad de priorización del algoritmo:

* **Efecto Sinergístico de las Variables:** El paso de 6 a 117 verdaderos positivos (TP) en el presupuesto de 800 alertas demuestra que el modelo `HistGradientBoosting` utiliza la métrica de rareza estadística para corregir el sesgo de saturación que afectaba al clasificador supervisado original. El algoritmo aprovecha la variable adicional para trazar fronteras de decisión que diferencian las señales reales del ruido de fondo en regiones de baja densidad local.
* **Eficiencia Operativa Basal:** La métrica *Precision@K* se eleva al 14.62%, lo que reduce el volumen de falsos positivos en el segmento prioritario. Esto resulta en un *Lift@K* de `14.625`, transformando un desempeño inferior al azar en una herramienta analítica capaz de multiplicar por 14 veces la eficiencia de inspección del laboratorio.
* **Reordenamiento Global del Espacio (AP = 0.0809):** El incremento del Average Precision global confirma que la inyección de la característica de distancia no lineal reestructura de forma general la asignación de probabilidades, separando de manera más efectiva la clase minoritaria a lo largo de todo el conjunto de prueba.

Integrar el enfoque no supervisado como un predictor del modelo supervisado es una estrategia altamente efectiva. Resuelve el problema del solapamiento geométrico al permitir que el árbol de decisión filtre las alertas utilizando simultáneamente la firma de la señal y la densidad del espacio físico.


### 5.2. Estrategia: Sistema de Alertas por Regla de Consenso Jerárquico

Esta estrategia combina los enfoques supervisado y no supervisado mediante una regla de decisión lógica basada en la coincidencia de sus alertas independientes. En lugar de reentrenar un modelo, las 800 alertas de HGB y las 800 alertas de LOF del conjunto de prueba se unifican en una lista de prioridades jerárquicas. 

El presupuesto estricto de 800 alertas totales se asigna de la siguiente manera: primero se seleccionan las observaciones donde ambos modelos coinciden (Prioridad 1) y el espacio restante se completa con los eventos detectados exclusivamente por el modelo no supervisado LOF (Prioridad 2), debido a su mayor precisión individual en el segmento prioritario.


```python
# Recuperar las máscaras booleanas (True/False) calculadas en la sección 4
mask_hgb, _ = alert_mask_scores(prob_test_final, 0.01)
mask_lof, _ = alert_mask_scores(scores_lof_test, 0.01)

# Solo LOF = Prioridad 2
# Solo HGB = Prioridad 1
# Ninguno = Prioridad 0
puntuacion_consenso = np.zeros(len(y_test))
puntuacion_consenso[mask_hgb & mask_lof] = 3
puntuacion_consenso[mask_lof & ~mask_hgb] = 2
puntuacion_consenso[mask_hgb & ~mask_lof] = 1

indices_ordenados = np.argsort(puntuacion_consenso)[::-1]
indices_priorizados = indices_ordenados[:PRESUPUESTO_TEST]

# Máscara final de alertas activadas por el Consenso
mask_consenso = np.zeros(len(y_test), dtype=bool)
mask_consenso[indices_priorizados] = True

tn, fp, fn, tp = confusion_matrix(y_test, mask_consenso).ravel()
precision_k = tp / PRESUPUESTO_TEST
recall_k = tp / int(y_test.sum())
lift_k = precision_k / y_test.mean()

res_consenso = {
    "Modelo": "Regla de Consenso (HGB + LOF)",
    "Alertas": int(mask_consenso.sum()),
    "TP": int(tp),
    "FP": int(fp),
    "Precision@K": precision_k,
    "Recall@K": recall_k,
    "Lift@K": lift_k,
    "AP": np.nan  
}

df_integracion = pd.DataFrame([
    df_comparativa_alertas[df_comparativa_alertas["Modelo"] == "Supervisado (HGB)"].iloc[0].to_dict(),
    df_comparativa_alertas[df_comparativa_alertas["Modelo"] == "No Supervisado (LOF)"].iloc[0].to_dict(),
    res_integrado,  
    res_consenso 
])

display(df_integracion[["Modelo", "Alertas", "TP", "FP", "Precision@K", "Recall@K", "Lift@K", "AP"]].round(4))
```

### 5.3. Comparación y Determinación de la Capacidad de Detección y Priorización

El contraste cuantitativo entre las estrategias integradas y los modelos individuales bajo el presupuesto estricto de 800 alertas permite determinar con precisión el impacto en el rendimiento:

* **Desempeño de la Estrategia Inyección de Características:** Al incluir el score de densidad local de LOF como un predictor más dentro de HGB, el modelo pasa de 6 a 117 verdaderos positivos (TP). Esto incrementa la precisión en el segmento prioritario (*Precision@K*) del 0.75% al 14.62% y el *Lift@K* de 0.750 a 14.625. Además, el Average Precision (AP) global se eleva a 0.0809, lo que demuestra un reordenamiento efectivo de todo el conjunto de prueba.
* **Desempeño de la Estrategia Regla de Consenso:** La combinación lógica por niveles de prioridad obtiene 16 verdaderos positivos (TP) y un *Lift@K* de 2.000. Aunque duplica la eficiencia del azar y supera al modelo HGB original, no aporta información nueva al algoritmo, quedando limitada por la baja tasa de aciertos individuales de las alertas base.

#### Mejora en la Detección y Priorización:

1. **Mejora en la Priorización (Top-K Alertas):** La Estrategia de inyección mejora de forma drástica la priorización de alertas. Resuelve la saturación de falsos positivos que afectaba al modelo supervisado individual en los rangos de máxima probabilidad, logrando concentrar la mayor densidad de señales reales en el primer 1.00% de la capacidad de revisión del laboratorio.
2. **Mejora en la Detección General:** El incremento sustancial del AP global confirma que la estrategia de inyección no solo optimiza la punta de la lista, sino que mejora la detección general del Bosón de Higgs. El modelo utiliza la métrica de distancia no lineal para resolver el solapamiento geométrico de las variables físicas originales.

La estrategia de inyección de características es la opción óptima para el despliegue del proyecto. Demuestra que el enfoque no supervisado es ineficiente de forma individual debido a la geometría del problema, pero funciona como un descriptor de densidad local que potencia y limpia la capacidad predictiva del clasificador supervisado.
